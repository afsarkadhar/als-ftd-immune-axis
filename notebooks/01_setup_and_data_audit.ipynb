```r
library(GEOquery)
library(tidyverse)
gse <- getGEO("GSE153960", GSEMatrix = TRUE, getGPL = FALSE)
pheno_list <- lapply(gse, function(x) pData(phenoData(x)))
get_col <- function(p, nm) if (nm %in% colnames(p)) p[[nm]] else NA_character_
meta <- bind_rows(lapply(names(pheno_list), function(n) {
  p <- pheno_list[[n]]
  tibble(gsm = rownames(p), series = n,
         title = get_col(p, "title"),
         group = get_col(p, "group:ch1"),
         tissue = get_col(p, "tissue:ch1"),
         libprep = get_col(p, "library preparation method:ch1"),
         subject = get_col(p, "subject id:ch1"),
         platform = get_col(p, "platform_id"),
         instrument = get_col(p, "instrument_model"))
}))
dim(meta)
meta %>% count(series, platform, instrument)
meta %>% count(group, sort = TRUE)
meta %>% count(tissue, sort = TRUE)
meta %>% count(libprep)
n_distinct(meta$subject)
```

```r
library(dplyr)
library(tidyr)
meta %>% dplyr::count(series, platform, instrument)
meta %>% dplyr::count(group, sort = TRUE)
meta %>% dplyr::count(tissue, sort = TRUE)
meta %>% dplyr::count(libprep)
n_distinct(meta$subject)
```
